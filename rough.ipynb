```python
| **Memory Type**                       | **What It Does**                                                                                                                                                                         | **When to Use It**                                                                                                                                            |
| ------------------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| 🗒 **ConversationBufferMemory**       | Stores the entire conversation word-for-word. Every time the user or AI speaks, it adds the new message to a list and sends the full chat history back to the model on the next prompt.  | Use this when building **simple chatbots** or **short assistants** where keeping full conversation context is important (e.g., “Remember my last question?”). |
| 🔁 **ConversationBufferWindowMemory** | Works like BufferMemory, but only remembers the *last few exchanges* instead of everything. You can decide how many previous turns to keep (e.g., last 3 messages).                      | Best for **long conversations** where full history would exceed token limits. Keeps the discussion recent and relevant.                                       |
| 🧮 **ConversationTokenBufferMemory**  | Instead of counting messages, it tracks **token usage**. When the total tokens reach a set limit, it drops older messages automatically.                                                 | Perfect when using **OpenAI GPT models** or **token-sensitive APIs** — ensures memory never exceeds the model’s max context window.                           |
| 🧾 **ConversationSummaryMemory**      | Summarizes earlier parts of the chat into short text, keeping only key ideas. It combines a summary + recent messages so the model still “remembers” context efficiently.                | Ideal for **very long chats** (like customer support or tutoring) — keeps the memory short, affordable, and context-aware.                                    |
| 📚 **VectorStoreRetrieverMemory**     | Converts each chat message into an **embedding** (numerical vector) and stores it in a vector database. When the user says something, it retrieves *semantically related past messages*. | Use this for **long-term or knowledge-based memory**, like personal assistants, FAQ bots, or systems that need to “recall” facts mentioned days ago.          |
| 🧑‍💼 **EntityMemory**                | Tracks information about **specific entities** (like people, products, or projects). For example, it can remember “Alice = customer from New York” or “Project X = delayed by 2 weeks.”  | Great for **personalized assistants** that need to remember user details, client data, or recurring topics over time.                                         |
```
